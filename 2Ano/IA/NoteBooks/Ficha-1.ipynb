```python
import pandas as pd
import numpy as np
df = pd.read_csv("../DataBases/db-Streaming.csv")
```

#### Cleaning: How many customers are missing the age? Fill in these values with the median age.

```python
missingValues = df.isnull().sum()
print(missingValues)

df["idade"] = df["idade"].fillna(df["idade"].median())
df["mensalidade"] = df["mensalidade"].fillna(df["mensalidade"].median())

df
```

#### Cardinality: How many unique values are there in column suporte_tecnico? And in the target?

```python
print(f"Unique Values of suporte_tecnico: {df["suporte_tecnico"].unique()}")
print(f"Unique Values of target: {df["target"].unique()}")
```

#### Frequency: What percentage of customers have "Canceled" the service? (Tip: use value_counts(normalize=True)).

```python
canceledPercentage = df["target"].value_counts(normalize=True) * 100
print(canceledPercentage)
```
